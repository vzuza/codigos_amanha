```python
from google.colab import drive
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Caminho dos dados no drive (IBGE)
drive.mount('/content/drive')

caminho_alfabetizacao ='/content/drive/MyDrive/Codigos_Projeto/IBGE/alfabetizacao_BR.xlsx'
caminho_basico = '/content/drive/MyDrive/Codigos_Projeto/IBGE/basico_BR.xlsx'
caminho_domicilio2= '/content/drive/MyDrive/Codigos_Projeto/IBGE/caracteristicas_domicilio2_BR.xlsx'
caminho_domicilio3 = '/content/drive/MyDrive/Codigos_Projeto/IBGE/caracteristicas_domicilio3_BR.xlsx'
caminho_demografia = '/content/drive/MyDrive/Codigos_Projeto/IBGE/demografia_BR.xlsx'
```

```python
#Visualização dos dados brutos IBGE - BR

df_alfabetizacao_BR = pd.read_excel(caminho_alfabetizacao)
df_basico_BR = pd.read_excel(caminho_basico)
df_domicilio2_BR = pd.read_excel(caminho_domicilio2)
df_domicilio3_BR = pd.read_excel(caminho_domicilio3)
df_demografia_BR = pd.read_excel(caminho_demografia)

print(df_alfabetizacao_BR.head())
print(df_basico_BR.head())
print(df_domicilio2_BR.head())
print(df_domicilio3_BR.head())
print(df_demografia_BR.head())
```

```python
#Limpando os dados do IBGE BR para IBGE PR

df_alfabetizacao_PR = df_alfabetizacao_BR[
    df_alfabetizacao_BR['CD_MUN'].astype(str).str.startswith('41')
].copy()
print(f'A base de dados possui {df_alfabetizacao_PR.shape[0]} linhas e {df_alfabetizacao_PR.shape[1]} colunas')

df_basico_PR = df_basico_BR[
    df_basico_BR['CD_MUN'].astype(str).str.startswith('41')
].copy()
print(f'A base de dados possui {df_basico_PR.shape[0]} linhas e {df_basico_PR.shape[1]} colunas')

df_domicilio2_PR = df_domicilio2_BR[
    df_domicilio2_BR['CD_MUN'].astype(str).str.startswith('41')
].copy()
print(f'A base de dados possui {df_domicilio2_PR.shape[0]} linhas e {df_domicilio2_PR.shape[1]} colunas')

df_domicilio3_PR = df_domicilio3_BR[
    df_domicilio3_BR['CD_MUN'].astype(str).str.startswith('41')
    ].copy()
print(f'A base de dados possui {df_domicilio3_PR.shape[0]} linhas e {df_domicilio3_PR.shape[1]} colunas')

df_demografia_PR = df_demografia_BR[
    df_demografia_BR['CD_MUN'].astype(str).str.startswith('41')
].copy()
print(f'A base de dados possui {df_demografia_PR.shape[0]} linhas e {df_demografia_PR.shape[1]} colunas')
```
