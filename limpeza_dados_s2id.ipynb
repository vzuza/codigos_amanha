```python
import pandas as pd
import os
import re
from glob import glob
```

```python
from google.colab import files

arquivos = files.upload()
```

```python

caminhos = glob('/content/dados-*.xls') + \
           glob('/content/dados-*.xlsx')

print("Arquivos encontrados:", len(caminhos))

lista_dfs = []

for arquivo in caminhos:

    ano = int(re.search(r'(\d{4})', os.path.basename(arquivo)).group(1))

    df = pd.read_excel(
        arquivo,
        skiprows=4,
        dtype=str
    )

    df.columns = df.columns.astype(str).str.strip()

    df = df.dropna(how='all')

    df['ANO_BASE'] = ano

    df['ARQUIVO_ORIGEM'] = os.path.basename(arquivo)

    lista_dfs.append(df)

df_s2id = pd.concat(
    lista_dfs,
    ignore_index=True,
    sort=False
)

print("Total de registros:", len(df_s2id))
print("Total de colunas:", len(df_s2id.columns))
```

```python
display(df_s2id.head())

print(df_s2id.columns.tolist())

df_s2id.groupby('ANO_BASE').size().reset_index(
    name='TOTAL_REGISTROS'
)
```

```python

df_s2id.to_csv(
    '/content/S2ID_PR_2015_2025.csv',
    index=False,
    encoding='utf-8-sig'
)

df_s2id.to_excel(
    '/content/S2ID_PR_2015_2025.xlsx',
    index=False
)

files.download('/content/S2ID_PR_2015_2025.csv')
```
