```python

from google.colab import drive
from IPython.display import display, IFrame
from html import escape
import os
import shutil
import zipfile

drive.mount('/content/drive')

PASTA = '/content/drive/MyDrive/Codigos_Projeto'
PASTA_RESULTADOS = os.path.join(PASTA, 'Resultados')

ARQUIVO_DASHBOARD = os.path.join(PASTA, 'dashboard_ippdn.html')
```

```python


mapas = []
graficos = []

for arquivo in os.listdir(PASTA):

    caminho = os.path.join(PASTA, arquivo)

    if arquivo.lower().endswith('.html') and 'mapa' in arquivo.lower():
        mapas.append(caminho)

    elif arquivo.lower().endswith('.png') and arquivo.lower().startswith('grafico_'):
        graficos.append(caminho)

ordem_mapas = [
    'Mapa_IPPDN_PR.html',
    'Mapa_Vulnerabilidade_PR.html',
    'Mapa_Frequencia_PR.html',
    'Mapa_Impacto_PR.html'
]

mapas.sort(
    key=lambda caminho: ordem_mapas.index(os.path.basename(caminho))
    if os.path.basename(caminho) in ordem_mapas
    else len(ordem_mapas)
)

graficos.sort()

print('Mapas encontrados:', len(mapas))
for caminho in mapas:
    print(os.path.basename(caminho))

print('\nGráficos encontrados:', len(graficos))
for caminho in graficos:
    print(os.path.basename(caminho))
```

```python

titulos_mapas = {
    'Mapa_IPPDN_PR.html':
        'Índice de Prioridade Preventiva de Desastres Naturais',

    'Mapa_Vulnerabilidade_PR.html':
        'Vulnerabilidade socioambiental',

    'Mapa_Frequencia_PR.html':
        'Frequência de desastres',

    'Mapa_Impacto_PR.html':
        'Impacto dos desastres'
}

titulos_graficos = {
    'grafico_1.png':
        'Distribuição do IPPDN por grau de prioridade',

    'grafico_2.png':
        'Distribuição dos valores do IPPDN',

    'grafico_3.png':
        'Frequência × Impacto',

    'grafico_4.png':
        'Vulnerabilidade × Frequência',

    'grafico_5.png':
        'Vulnerabilidade × Impacto'
}
```

```python

secoes_mapas = ''

for caminho in mapas:
    arquivo = os.path.basename(caminho)
    titulo = titulos_mapas.get(arquivo, arquivo)

    secoes_mapas += f"""
    <section class="card mapa-card">
        <h3>{escape(titulo)}</h3>
        <iframe
            src="{escape(arquivo)}"
            loading="lazy"
            title="{escape(titulo)}">
        </iframe>
    </section>
    """
```

```python

secoes_graficos = ''

for caminho in graficos:
    arquivo = os.path.basename(caminho)
    titulo = titulos_graficos.get(arquivo, arquivo)

    secoes_graficos += f"""
    <section class="card grafico-card">
        <h3>{escape(titulo)}</h3>
        <img
            src="{escape(arquivo)}"
            alt="{escape(titulo)}"
            loading="lazy">
    </section>
    """
```
